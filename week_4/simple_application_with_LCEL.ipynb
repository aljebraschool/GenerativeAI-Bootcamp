### Using open source model with Groq API key

```python
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
groq_api_key = os.getenv("GROQ_API_KEY")
openai_api_key = os.getenv("OEPNAI_API_KEY")
langchain_api_key = os.getenv("LANGCHAIN_API_KEY")
langchain_project = os.getenv("LANGCHAIN_PROJECT")
langchain_tracing = os.getenv("LANGCHAIN_TRACING_V2", 'true')
```

```python
from langchain_groq import ChatGroq
```

```python
model = ChatGroq(model = 'deepseek-r1-distill-llama-70b', groq_api_key = groq_api_key) # gemma2-9b-it, deepseek-r1-distill-llama-70b

model
```

```python
from langchain_core.messages import HumanMessage, SystemMessage
from IPython.display import display, Markdown
```

```python
messages = [
    SystemMessage(content = "Translate the following from english to French"),
    HumanMessage(content = "Hello, How are you?")
]
```

```python
response = model.invoke(messages)

response 
```

```python
display(Markdown(f"{response}"))
```

```python
#print out only the content of the AIMessage using output parser
from langchain_core.output_parsers import StrOutputParser
```

```python
output_parser = StrOutputParser()

output_parser.invoke(response)
```

### We can use LCEL (Langchain Expression Language) to chain the components (llm model and output parser) given above

```python
chain = model | output_parser

display(Markdown(chain.invoke(messages)))
```

### Using prompt template 

```python
from langchain.prompts import ChatPromptTemplate
```

```python
template_string = """
As a model that translates from one language to another, help user translate this {text} into the given {language}:
text : {text}
langauge : {language}
"""

prompt_template = ChatPromptTemplate.from_template(template_string)
```

```python
prompt = prompt_template.format_prompt(text = "Hello, how are you?", language = "French")
```

```python
display(Markdown(f"{model.invoke(prompt)}"))
```

```python
from langchain_core.messages import HumanMessage
```

```python
chain = prompt_template| model | output_parser

# Invoke the chain with a dictionary of inputs
response = chain.invoke({
    "text": "hello, how are you?",
    "language": "french"
})

display(Markdown(response))
```
