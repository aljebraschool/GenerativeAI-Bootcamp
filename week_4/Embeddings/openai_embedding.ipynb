### Now implement vector Embedding

```python
from langchain_openai import OpenAIEmbeddings
```

```python
#create the object of the embedding class 
embeddings = OpenAIEmbeddings(model = 'text-embedding-3-large')
```

```python
embeddings
```

```python
text = "This is an embedding"

embeddings.embed_query(text)
```

### Implementing chunking of document

```python
#import library 
from langchain_community.document_loaders import PyPDFLoader
```

```python
#initialize the loader
loader = PyPDFLoader('/home/aljebra/GenerativeAI_Projects/python-codes/NLP/langchain/Dataingestion/Ridwan Recommendation (1).pdf')
```

```python
pdf_file = loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text_splitters = RecursiveCharacterTextSplitter(chunk_size = 100, chunk_overlap = 20)

final_doc = text_splitters.split_documents(pdf_file)
```

### Now convert this document to vector embedding and store it in vector store

```python
from langchain_community.vectorstores import Chroma
```

```python
vector_store_db = Chroma.from_documents(final_doc, embeddings)
```

```python
vector_store_db
```

### How to query from this vector store 

```python
query = "His efforts and unique approach eventually paid off a"
```

```python
vector_store_db.similarity_search(query)
```
