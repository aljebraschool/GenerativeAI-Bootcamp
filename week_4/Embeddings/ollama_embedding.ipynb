```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embedding = (
    OllamaEmbeddings(model='llama3') #by default it uses llama2. You can put another model using the parameter using (model = 'llama3')
)
```

```python
embedding
```

```python
#now use this to embed text
vector = embedding.embed_documents(
    [
        "I love coding in Python",
        "Programming is th language of computer"
    ]
)
```

```python
vector[1]
```

```python
#let use the embedding to ask question 
embedding.embed_query("what is the language of computer") # this will give vector that is very close to the second second generated by the embedding
```
