# Artificial Neural Network

image.png

### A Neural Network is a computational model inspired by the human brain that learns patterns from data by adjusting connections between simple processing units called neurons.

### Think of it as a mathematical function that can learn to map input data (like images or text) to desired outputs (like labels or predictions) by finding complex patterns through examples.

## Simple Analogy

image.png

### Imagine teaching a child to recognize animals. You show many pictures, say "dog" when you see a dog, and correct mistakes. The child's brain strengthens connections for correct features. A neural network does the same digitally—adjusting internal "knobs" (weights) when it makes errors until it learns.

## Key Point

### Neural networks are composed of layers of interconnected neurons that transform input data step-by-step. They learn by adjusting connection strengths (weights) based on errors, using a process called backpropagation.

## Basic Structure

image.png

1. **Input Layer**: Receives the data (e.g., pixels of an image).

2. **Hidden Layers**: Intermediate layers that detect patterns (edges, shapes, concepts).

3. **Output Layer**: Produces the final result (e.g., "cat" or "dog").

## Why it's used

### Neural networks excel at finding complex patterns in unstructured data (images, sound, text). They're the foundation of modern AI and are used for:

- Image Recognition: Face detection, medical imaging.

- Natural Language Processing: Translation, chatbots.

- Predictive Analytics: Stock market forecasts, recommendation systems.

- Autonomous Systems: Self-driving cars, game-playing AI.



> Note: Deep Learning refers to neural networks with many hidden layers, enabling them to learn hierarchical representations—from simple features to abstract concepts.

# Forward Propagation

image.png

### Forward Propagation is the process of passing input data through a neural network layer-by-layer to compute an output prediction.

### Think of it as pushing data through a factory assembly line: raw materials enter at one end, get processed at each station (layer), and a finished product emerges at the other end.

```python

Input → [Layer 1: Weights×Input + Bias → Activation] → [Layer 2: Same process] → ... → Final Output

## Key Point

### Forward propagation is just calculation, not learning. It uses the current network parameters (weights and biases) to make a prediction, without changing them.

## Why It Matters



- It's the "forward pass" that generates predictions for evaluation

- Essential for both making predictions and for the subsequent backpropagation phase (where learning happens by adjusting weights based on errors)

- Every neural network inference (using the model) involves forward propagation

# Backward Propagation (Backpropagation)

### Backward Propagation is the learning process in neural networks where the error from the output is passed backward through the network to adjust the weights, making future predictions more accurate.

### Think of it as tracing your steps backward through a maze after reaching a dead end, noting where you took wrong turns so you can adjust your path next time.

```python

Input → Forward Pass → Output → Compare to Truth → Calculate Error →  Backward Pass → Adjust Weights → Repeat

## Key Point


### Backpropagation is the "learning" phase. It's what allows neural networks to improve from experience by making systematic adjustments to all the connection strengths (weights).

## Why It Matters


- It's the fundamental algorithm that enables training of deep neural networks

- Allows networks to learn complex patterns by making thousands/millions of tiny adjustments

- Makes deep learning scalable by efficiently calculating error contributions for all weights

# Regularization

image.png

### Regularization is a set of techniques used in machine learning to prevent a model from becoming too complex and memorizing the training data (overfitting), so it performs better on new, unseen data.

### Think of it as adding constraints or penalties to a student who memorizes textbook examples perfectly but fails on new problems—it encourages learning the underlying concepts instead.

## Common Regularization Techniques:


### 1. L1 Regularization (Lasso)


- Adds penalty proportional to the absolute value of weights

- Effect: Can drive some weights to exactly zero → performs feature selection

- Formula: Loss = Original Loss + λ × sum(|weights|)

### 2. L2 Regularization (Ridge)


- Adds penalty proportional to the square of weights

- Effect: Shrinks all weights but rarely sets them to zero

- Formula: Loss = Original Loss + λ × sum(weights²)



### 3. Dropout


- Randomly "drops out" (temporarily removes) neurons during training

- Effect: Prevents co-adaptation of neurons, forces network to be robust

- Like training multiple thinned networks and averaging them

### 4. Early Stopping


- Stop training when performance on validation set stops improving

- Effect: Prevents the model from learning noise in the training data

# Key Point


### Regularization intentionally reduces training accuracy slightly to increase generalization to new data. It's the "less is more" principle in machine learning.

> Note: The regularization strength (λ) is a hyperparameter—too little and you overfit, too much and you underfit. It's about finding the right balance between simplicity and accuracy.
