# Long Short-Term Memory (LSTM)

image.png

### LSTM is a specialized type of recurrent neural network designed to effectively capture and remember long-term dependencies in sequential data, overcoming the limitations of standard RNNs.



### Think of it as a smart memory system with controlled gates that decide what information to keep, what to update, and what to forget—like a librarian who selectively preserves important books while removing outdated ones.



## Simple Analogy

image.png

### Imagine you're reading a mystery novel:

- Standard RNN: Remembers only the last few pages
- LSTM: Keeps track of clues from early chapters, character backgrounds, and plot twists while ignoring irrelevant details like weather descriptions

## Core Concept: The Memory Cell with Three Gates

### LSTMs have a cell state (long-term memory) that runs through the entire sequence, plus three regulatory gates:



## 1. Forget Gate: Decides what information to discard

- Looks at current input + previous hidden state
- Outputs values between 0 (forget completely) and 1 (remember completely)
- Formula: f_t = σ(W_f · [h_(t-1), x_t] + b_f)

## 2. Input/Learn Gate: Decides what new information to store

Two parts:

a) Input gate layer: Decides which values to update (0-1)
i_t = σ(W_i · [h_(t-1), x_t] + b_i)

b) Candidate layer: Creates new candidate values to add
C̃_t = tanh(W_C · [h_(t-1), x_t] + b_C)

## 3. Output Gate: Decides what to output based on cell state

- Controls what parts of the cell state are exposed as output
- Formula: o_t = σ(W_o · [h_(t-1), x_t] + b_o)



### The Complete LSTM Process at Each Step:

### Step 1: Forget Unnecessary Information



```
What to forget = σ(Weights_f · [Previous Hidden State, Current Input] + Bias_f)

New Cell State = Old Cell State × What to forget
```



### Step 2: Learn New Information



```
What to learn = σ(Weights_i · [Previous Hidden State, Current Input] + Bias_i)
Candidate values = tanh(Weights_C · [Previous Hidden State, Current Input] + Bias_C)

New Cell State = New Cell State + (What to learn × Candidate values)
```



### Step 3: Generate Output



```
Output gate = σ(Weights_o · [Previous Hidden State, Current Input] + Bias_o)

Hidden State (output) = Output gate × tanh(New Cell State)
```



## Why This Architecture Works:


### 1. Controlled Information Flow:

- The cell state is the "conveyor belt" running straight through with minor linear interactions

- Gates allow precise control of information flow without constant overwriting

## 2. Solves Vanishing Gradient Problem:

- The cell state provides a "highway" for gradients to flow through time

- Forget/input gates use sigmoid (0-1 range), preventing gradient explosion


## 3. Selective Memory:

- Can remember information for hundreds of time steps when needed

- Can quickly discard irrelevant information

## Practical Example: Language Translation

### Translating English to French:

**Input: "The beautiful mountain that we visited last summer was covered in snow"**

### LSTM Operations:

1. Forget Gate: Discards article "the" after it's processed
2. Input Gate: Stores "beautiful" to modify upcoming noun
3. Remember Gate: Keeps "mountain" as subject until verb appears
4. Output Gate: Releases "la belle montagne" in French when structure is complete

> Key Memory: Remembers "mountain" across 7 words until reaching the verb "was covered"

## Why LSTMs Are Used:

### They excel where long-term context matters:

- Machine Translation: Remembering subject-verb agreement across long sentences

- Speech Recognition: Connecting phonemes to words across time

- Time Series Prediction: Stock market trends over months/years

- Medical Records: Tracking patient history over years

> **Note: While Transformers have surpassed LSTMs for many NLP tasks, LSTMs remain valuable for: real-time streaming data, computational efficiency, and situations with strict sequential dependencies.**

## Key Insight:


> LSTMs don't just have better memory—they have smarter memory management. The gates learn through training what information is important to keep, update, or discard for each specific task.
